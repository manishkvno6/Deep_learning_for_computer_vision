```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt 
```

```python
#reading as greyscale image
image_path = r'D:\CVDL\images\Gemini_Generated_Image_22ypm122ypm122yp.png'
img = cv.imread(image_path, cv.IMREAD_GRAYSCALE)

```

```python
#resizing the image to 800x600
img = cv.resize(img, (800,600))
cv.imshow('Orignal image', img)
cv.waitKey(0)
```

```python
#storing the shape of the image in row and col variables
row, col = img.shape
```

```python
#making sobel horizontal kernel
sobel_hor_kernel = np.array([
    [1,2,1],
    [0,0,0],
    [-1,-2,-1]
])
```

```python
#Assigning kernal shape
k = sobel_hor_kernel.shape[0]
```

```python
output_matrix = np.zeros((row - k + 1, col - k + 1))
output_matrix.shape
```

```python
for i in range(1, row-1):
    for j in range(1,col-1):
        or_img_region = img[i - 1:i + 2,j-1:j + 2]
        output_matrix[i - 1,j - 1] = np.sum(or_img_region * sobel_hor_kernel)
```

```python
edg_det = output_matrix
cv.imshow('After_sobel', edg_det)
cv.waitKey(0)
```
